##What is Z-Ordering in Delta Lake?

Z-Ordering (ZORDER BY) is a technique in Databricks Delta Lake to co-locate related data in the same files.

It’s mainly used to improve query performance when filtering on specific columns (especially non-partitioned columns).

##Why We Need Z-Ordering

Even after partitioning, queries on non-partition columns may scan many files.

👉 Example:
If your table is partitioned by year, but queries often use WHERE movie_name = 'Inception',
partitioning doesn’t help much.

That’s where ZORDER helps — it clusters data on disk by the given column(s).<br>
##Syntax
OPTIMIZE <table_name>
ZORDER BY (column1 [, column2, ...]);
