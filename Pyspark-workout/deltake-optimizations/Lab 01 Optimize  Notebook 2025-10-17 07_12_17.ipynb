##What is OPTIMIZE in Databricks?

In Delta Lake, data is stored in multiple small Parquet files.
Over time, as new data is appended or updated, you may end up with hundreds or thousands of small files (called small file problem).
These many small files slow down query performance.

✅ The OPTIMIZE command helps by merging small files into fewer large files, improving:

Query speed

Scan efficiency

Data skipping

🔹 Syntax
OPTIMIZE delta.

```python
%sql
CREATE TABLE IF NOT EXISTS databricks_practice.inputdb.employees_opt_z_or (
  id INT,
  name STRING,
  department STRING,
  salary DECIMAL(10, 2),
  hire_date DATE
) USING DELTA;
```

```python
%sql
DESCRIBE DETAIL databricks_practice.inputdb.employees_opt_z_or;
```

```python
%sql
OPTIMIZE databricks_practice.inputdb.employees_opt_z_or ZORDER BY (department);
```
