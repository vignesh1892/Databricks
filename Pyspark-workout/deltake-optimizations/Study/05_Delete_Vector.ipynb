Deletion Vectors are a modern performance feature in Delta Lake (introduced in Delta Lake 2.3 and Databricks Runtime 12.1) designed to solve the problem of Write Amplification during DELETE, UPDATE, and MERGE operations.

1. The Problem: Write Amplification (The "Old" Way)
Before Deletion Vectors, Delta Lake used a Copy-on-Write strategy:

If you wanted to delete just 1 row inside a 100 MB Parquet file, Databricks had to read the entire 100 MB file, remove that one row in memory, and then write a brand new 99.9 MB Parquet file.
The Cost: This is extremely expensive in terms of I/O and compute, especially for small, frequent updates.
2. The Solution: Merge-on-Read (The "New" Way)
Deletion Vectors change this to a Merge-on-Read strategy:

Instead of rewriting the whole data file, Databricks simply creates a tiny "sidecar" file called a Deletion Vector.
The Mechanism: The Deletion Vector is essentially a bitmap that says: "In File A, ignore row numbers 5, 12, and 100."
The Benefit: The DELETE or UPDATE command finishes almost instantly because it only writes a few bytes (the bitmap) instead of megabytes of data.
