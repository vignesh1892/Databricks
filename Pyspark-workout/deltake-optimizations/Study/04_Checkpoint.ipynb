Checkpoints are a critical performance feature designed to manage the growth of the transaction log (_delta_log). Without checkpoints, Spark would have to read every single JSON file in the log directory to determine the current state of a table, which would become slower with every transaction.
**Types:**
- Classic Checkpoints (V1)
 ALTER TABLE table_name SET TBLPROPERTIES ('delta.checkpointInterval' = '20')
- V2 Checkpoints (Sidecars)
ALTER TABLE table_name SET TBLPROPERTIES ('delta.checkpointPolicy' = 'v2')
