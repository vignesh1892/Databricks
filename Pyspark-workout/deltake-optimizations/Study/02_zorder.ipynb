**What is Z-Ordering?**<br>
Physically co-locates related data across multiple columns on disk, using a multi-dimensional sorting approach (Z-order curve) to improve query performance, especially for filters and joins on high-cardinality columns like IDs or timestamps, by enabling aggressive data skipping.

**Syntax:**
Its always performed along with optimize.
OPTIMIZE table_name
ZORDER BY (column_x, column_y);

**When to use Z-Ordering**
- High Cardinality Columns (e.g., user_id, order_id, email).
- Frequent Filter Columns
- Join Keys

**Limitations**<br>
- Do not Z-Order by too many columns.
- Maintenance: Z-Ordering is not a "one and done" operation. When you append new data to a table, that new data is not Z-Ordered. You must re-run the OPTIMIZE ... ZORDER BY command periodically to organize the newly arrived data.
- Compute Cost: Because Z-Ordering involves a global sort of your data, it is more resource-intensive than a plain OPTIMIZE
