##What is CTAS in Databricks?

CTAS stands for Create Table As Select.
It’s a SQL command used to create a new table and populate it with the results of a query — all in one step.

✅ You can think of it as:

“Create a table + Insert the data from SELECT.”

🔹 Basic Syntax
CREATE TABLE <table_name>
USING <format>
AS
SELECT <columns>
FROM <source_table>
WHERE <condition>;

🔹 Key Points
| Feature                     | Description                                     |
| --------------------------- | ----------------------------------------------- |
| Purpose                     | Create and populate a new table in one command  |
| Output Table Type           | Managed or External (depending on location)     |
| Data Format                 | Delta, Parquet, CSV, JSON, etc.                 |
| Automatically infers schema | Yes                                             |
| Automatically writes data   | Yes                                             |
| Works with any SELECT       | Yes, including joins, filters, and aggregations |
