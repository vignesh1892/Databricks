```python
from pyspark.sql.functions import max
from pyspark.sql.types import *
cust_schema = StructType([StructField('custid', StringType(), True), StructField('fname', StringType(), True), StructField('lname', StringType(), True), StructField('age', IntegerType(), True), StructField('profession', StringType(), True)])
profname=dbutils.widgets.get("prof")
df= spark.read.format("csv").option("header","true").schema(cust_schema).load("/Volumes/databricks_practice/inputdb/customerdata/custs.txt")

df.write.mode("overwrite").saveAsTable("databricks_practice.outputdb.tblprof")

max_age= df.select(max("age")).collect()[0][0]

print(max_age)

##passing value to another task
dbutils.jobs.taskValues.set("set_max_age",max_age)
```
