```python
df=spark.table("databricks_practice.inputdb.employees_sc")
display(df)
```

```python
dbutils.widgets.text("dept","Sales")
```

```python
_dept=dbutils.widgets.get("dept")
df1=df.filter(df["department"]==_dept)
df1.display()
df1.write.mode("overwrite").saveAsTable(f"databricks_practice.outputputdb.employees_dept_{_dept}")
```

```python
rowcount=df1.count()
print(f"Data processed for dept {_dept} and row count is {rowcount}")
dbutils.jobs.taskValues.set("rowcount",rowcount)
dbutils.jobs.taskValues.set("deptname",_dept)
```
