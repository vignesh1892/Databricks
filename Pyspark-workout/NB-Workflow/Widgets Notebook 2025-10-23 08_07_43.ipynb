###🧠 What are Databricks Widgets?

Databricks Widgets allow you to add input controls (like dropdowns, text boxes, or multi-select lists) to your notebooks.
They make your notebooks interactive — ideal for parameterizing notebooks and creating dynamic dashboards or reports.

⚙️ Why Use Widgets

To pass parameters to your notebook at runtime

To create reusable notebooks for different inputs

To make notebooks interactive and user-friendly

🧩 Types of Databricks Widgets
| Widget Type     | Function                            | Example                                                                                     |
| --------------- | ----------------------------------- | ------------------------------------------------------------------------------------------- |
| **Text**        | Free-form text input                | `dbutils.widgets.text("input_name", "default_value", "Label")`                              |
| **Dropdown**    | Single value from a fixed list      | `dbutils.widgets.dropdown("day", "Mon", ["Mon", "Tue", "Wed"], "Select Day")`               |
| **Combobox**    | Dropdown + allows custom text entry | `dbutils.widgets.combobox("fruit", "Apple", ["Apple", "Banana", "Grapes"], "Select Fruit")` |
| **Multiselect** | Select multiple values              | `dbutils.widgets.multiselect("colors", "Red", ["Red", "Blue", "Green"], "Select Colors")`   |

##commands:
| Command                         | Purpose                |
| ------------------------------- | ---------------------- |
| `dbutils.widgets.text()`        | Create text widget     |
| `dbutils.widgets.dropdown()`    | Create dropdown widget |
| `dbutils.widgets.combobox()`    | Dropdown + free text   |
| `dbutils.widgets.multiselect()` | Select multiple items  |
| `dbutils.widgets.get()`         | Retrieve widget value  |
| `dbutils.widgets.remove()`      | Remove a widget        |
| `dbutils.widgets.removeAll()`   | Remove all widgets     |




```python
#Example
dbutils.widgets.text("region","NA","select region")
dbutils.widgets.dropdown("year", "2024", ["2022", "2023", "2024", "2025"], "Select Year")

print("wigets creeated")
```

```python
_region=dbutils.widgets.get("region")
_year=dbutils.widgets.get("year")

print(f"select region:{_region}")
print(f"select year:{_year}")
```
