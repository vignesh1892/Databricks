```python
df= spark.read.format("csv").option("header","True").load("/Volumes/databricks_practice/inputdb/customerdata/custs.txt")
df.show(3)
```

```python
prof_array=[row["profession"] for row in df.select("profession").distinct().collect() if row["profession"] is not None]
print(prof_array)
# Optional: handle case when the list might be empty
default_value = prof_array[0] if len(prof_array) > 0 else "NA"

# Create a Databricks dropdown widget
dbutils.widgets.dropdown("profession_drop", default_value, prof_array, "Select Profession")
dbutils.widgets.combobox("profession_combo", default_value, prof_array, "Select Profession")
dbutils.widgets.multiselect("profession_multi", default_value, prof_array, "Select Profession")
```

```python
_prof=dbutils.widgets.get("profession")
if _prof.strip() != "":
    df.filter(df.profession==_prof).display()
else:
    print("Enter the profession")
```
