```python
from pyspark.sql.functions import lit
# Try to get risk flag from high risk task
risk_flag = dbutils.jobs.taskValues.get(
    taskKey="04_task_high_risk",
    key="risk_flag",
    default="NA",
    debugValue="NA"
)

# If not found, get from low risk task
if risk_flag == "NA":
    risk_flag = dbutils.jobs.taskValues.get(
        taskKey="04_task_low_risk",
        key="risk_flag",
        default="NA",
        debugValue="NA"
    )

# Read the existing table
loan_df = spark.table("databricks_practice.outputdb.tbl_loandata")

# Add the risk category column
final_df = loan_df.withColumn("risk_category", lit(risk_flag))

# Write results to a target table
final_df.write.mode("append").saveAsTable("databricks_practice.outputdb.tbl_loandata_risk")

print(f"Final Risk Evaluation results: {risk_flag}")
```
