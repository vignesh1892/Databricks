```python
dbutils.widgets.text("profession","Teacher","Select Profession")
dbutils.widgets.text("age","55","Select age")

_age=dbutils.widgets.get("age")

_profession=dbutils.widgets.get("profession")
```

```python
v_count=0
df= spark.read.format("csv").option("header","True").load("/Volumes/databricks_practice/inputdb/customerdata/custs.txt")
if _profession.strip() != "":
    df_filter=df.filter((df.profession==_profession) | (df.age == int(_age)))
    df_filter.display()
else:
    print("Enter the profession")
v_count =df_filter.count()
```

```python
#exit code (retrun value that should be in string format)
dbutils.notebook.exit(str(v_count))
```
