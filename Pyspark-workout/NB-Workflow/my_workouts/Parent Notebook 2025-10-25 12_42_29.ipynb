```python
v_region="20"
region_list = dbutils.jobs.taskValues.get("03_get_regions", "region_list", debugValue="50")

print(f"Region value received from previous task: {region_list}")
for i in region_list:
    region_df = spark.sql(
        f"""
        SELECT * 
        FROM databricks_practice.inputdb.tbl_country_population 
        WHERE REGION_ID = {i}
        """
    )
    region_df.show()
    _count = region_df.count()
    if _count > 0:
        region_df.write.format("delta") \
            .partitionBy("SUB_REGION_ID") \
            .mode("ignore") \
            .saveAsTable(f"databricks_practice.outputdb.tbl_region_{i}_population")
```
