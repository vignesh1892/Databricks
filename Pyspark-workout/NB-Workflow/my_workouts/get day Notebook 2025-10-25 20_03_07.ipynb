```python
input_curr_tp = dbutils.widgets.get("input_current_timestamp")
print(input_curr_tp)
```

```python
from pyspark.sql.functions import to_timestamp,date_format

_user_input_day = spark.sql(
    f"""SELECT date_format(
            try_to_timestamp('{input_curr_tp}', "yyyy-MM-dd'T'HH:mm:ss.SSS"),
            "E"
        )"""
).collect()[0][0]
print(_user_input_day)
if _user_input_day is not None:
    _user_input_day = _user_input_day.upper()
    print(_user_input_day)
else:
    print("Invalid or empty timestamp input.")
```

```python
dbutils.jobs.taskValues.set("input_current_timestamp",_user_input_day)
```
