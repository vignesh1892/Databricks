```python
from pyspark.sql.types import StructType, StructField, StringType,IntegerType
from pyspark.sql.functions import col

cust_sch=StructType([StructField('custid', IntegerType(), True), StructField('fname', StringType(), True), StructField('lname', StringType(), True), StructField('age', IntegerType(), True), StructField('profession', StringType(), True)])
df= spark.read.format("csv").option("header",True).schema(cust_sch).load("/Volumes/databricks_practice/inputdb/customerdata/custs.txt")
print(df.schema)

output_path="dbfs:/Volumes/databricks_practice/outputdb/customerdata/delta"
df.write.format("delta").mode("overwrite").partitionBy("profession","age").save(output_path)

df.write.format("delta").mode("overwrite").partitionBy("profession","age").saveAsTable("databricks_practice.outputdb.tblcustomer")

df.orderBy("profession","age").show()

df.sort("profession","age").show()

df.select("profession").distinct().display()

df.select("profession").distinct().count()



```
