
###Creating a Dataframe using createDataFrame API

```python
data = [
    {"emp_id": 101, "name": "Vignesh", "dept": "IT", "salary": 75000},
    {"emp_id": 102, "name": "Kumar", "dept": "Finance", "salary": 68000},
    {"emp_id": 103, "name": "Anjali", "dept": "HR", "salary": 72000}
]

df =spark.createDataFrame(data)


display(df)
```

```python
# from an row object
from pyspark.sql import Row
data =[Row(id=1,name="Vignesh",age=25),Row(id=2,name="Nandhu",age=25)]

df=spark.createDataFrame(data)
df.display()
```

```python
# empty DF
from pyspark.sql.types import StructType, StructField, StringType, IntegerType

# Define schema
emp_schema = StructType([
    StructField("emp_id", IntegerType(), True),
    StructField("name", StringType(), True),
    StructField("dept", StringType(), True),
    StructField("salary", IntegerType(), True)
])

df = spark.createDataFrame([],schema=emp_schema)
df.display()
```
