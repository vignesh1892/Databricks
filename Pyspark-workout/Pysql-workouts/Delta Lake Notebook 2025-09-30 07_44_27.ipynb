####working with delta file format

```python
from pyspark.sql.types import StructType, StructField, StringType, IntegerType, ArrayType
from pyspark.sql.functions import from_json,col


df_schema = StructType([StructField("studid",IntegerType(),True)
                          ,StructField("studname",StringType(),True)
                          ,StructField("studsports",StringType(),True)])

df=spark.read.format("csv").option("header",True).schema(df_schema).load("/Volumes/databricks_practice/inputdb/customerdata/sports_with_json_colum.csv")
df.display()
df.printSchema()

json_schema = StructType([StructField("sportsname",StringType(),True)
                          ,StructField("year",IntegerType(),True)
                          ,StructField("sportcoach",StringType(),True)])

df2=df.withColumn("json_col",from_json(col("studsports"),json_schema))
df2.display()
ot_path="dbfs:/Volumes/databricks_practice/outputdb/studdata/studata/studs_delta"
df2.write.format("delta").mode("overwrite").save(ot_path)



                                                       
```

```python
#storing it as delta table
df2.write.format("delta").saveAsTable("databricks_practice.outputdb.tblsports")
```
