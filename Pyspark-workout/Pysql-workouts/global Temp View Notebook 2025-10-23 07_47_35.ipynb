###🧠 What is a Global Temporary View?

A Global Temporary View is a temporary table that is shared across all sessions and notebooks within the same Spark application.

Unlike a local temporary view, which is visible only to the current session,
a global temporary view is visible to all sessions using a shared database called global_temp.

| Feature           | Description                                                                                         |
| ----------------- | --------------------------------------------------------------------------------------------------- |
| **Scope**         | Visible across all notebooks/sessions within the same cluster                                       |
| **Database**      | Stored in the built-in `global_temp` database                                                       |
| **Persistence**   | Exists until the Spark application (cluster) stops                                                  |
| **Access Syntax** | Must use the prefix `global_temp.` before the view name                                             |
| **Use Case**      | When you want to share temporary data between different notebooks or jobs in the same Spark session |


🔍 Accessing a Global Temporary View

You must use the global_temp. prefix:

# Query from the same notebook or another notebook in the same cluster
spark.sql("SELECT * FROM global_temp.cricketers").show()

📋 Show All Global Temporary Views

To list all global temporary views:

SHOW TABLES IN global_temp;


This will display all views created in the global_temp namespace.

🧹 Dropping a Global Temporary View
DROP VIEW global_temp.cricketers;


###⚖️Difference Between Local and Global Temp Views
| Feature           | Local Temp View               | Global Temp View                                |
| ----------------- | ----------------------------- | ----------------------------------------------- |
| **Scope**         | Current notebook/session only | Across all notebooks/sessions in same Spark app |
| **Access Prefix** | Not required                  | `global_temp.` required                         |
| **Persistence**   | Ends when the notebook ends   | Ends when Spark application stops               |
| **Database Name** | None                          | `global_temp` (built-in)                        |


```python
from pyspark.sql.types import StructType, StructField, StringType, IntegerType, ArrayType
from pyspark.sql.functions import from_json,col


df_schema = StructType([StructField("studid",IntegerType(),True)
                          ,StructField("studname",StringType(),True)
                          ,StructField("studsports",StringType(),True)])

df=spark.read.format("csv").option("header",True).schema(df_schema).load("/Volumes/databricks_practice/inputdb/customerdata/sports_with_json_colum.csv")
df.display()
```

```python
df.createOrReplaceGlobalTempView("tblsports_gv")
df.createOrReplaceTempView("tblsports_v")
```
