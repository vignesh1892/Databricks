```python
#simple json
json_data = {
    "empid": 100,
    "empname": "Romin",
    "designation": "developer"
}
simple_json_df= spark.createDataFrame([json_data])
simple_json_df.display()

simple_json_df.select("empid","empname").show()
```

```python
# Read the JSON column from csv file
from pyspark.sql.types import StructType, StructField, StringType, IntegerType, ArrayType
from pyspark.sql.functions import from_json,col

multi_json = spark.read.option('multiLine', True).option("ignoreLeadingWhiteSpace", True).option("ignoreTrailingWhiteSpace", True).option('header', True).csv(
    "/Volumes/databricks_practice/inputdb/customerdata/sports_with_json_colum.csv"
)

json_schema = StructType([StructField("sportsname",StringType(),True)
                          ,StructField("year",IntegerType(),True)
                          ,StructField("sportcoach",StringType(),True)])

multi_json2=multi_json.withColumn("json_col",from_json(col("studsports"),json_schema))
multi_json2.display()

multi_json3=multi_json2.select("studid","studname","json_col.sportsname","json_col.year","json_col.sportcoach")
multi_json3.display()




```

```python
#Multi line Json
from pyspark.sql.functions import col
df1=spark.read.json("/Volumes/databricks_practice/inputdb/customerdata/nested_json.json",multiLine=True)
df1.display()

df1.printSchema()

df1.select(df1.id,df1.name,df1.details.age,df1.details.city,df1.hobbies[0].alias("skill1")).show()


df1.select("id","name","details.age","details.city",col("hobbies")[0].alias("skill1")).show()
```

```python
#explode option:
#Multi line Json
from pyspark.sql.functions import col,explode
df1=spark.read.json("/Volumes/databricks_practice/inputdb/customerdata/nested_json.json",multiLine=True)
df1.display()

df1.printSchema()

df1.withColumn("explode_col",explode("hobbies")).show()


df1.select("id","name","details.age","details.city",col("hobbies")[0].alias("skill1")).show()

```
