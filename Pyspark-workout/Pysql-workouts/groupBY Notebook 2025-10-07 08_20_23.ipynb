##groupBy() to aggerate df with one or more columns

```python
from pyspark.sql.types import StructType, StructField, StringType,IntegerType
from pyspark.sql.functions import col,concat,lit

cust_sch=StructType([StructField('custid', IntegerType(), True), StructField('fname', StringType(), True), StructField('lname', StringType(), True), StructField('age', IntegerType(), True), StructField('profession', StringType(), True)])
df= spark.read.format("csv").option("header",True).schema(cust_sch).load("/Volumes/databricks_practice/inputdb/customerdata/custs.txt")
print(df.schema)

df2=df.withColumn("region",lit("USA"))
df2.printSchema()
df2.groupBy(col("profession")).count().display()
```

#agg() => perfrom one or more aggreation on df
from pyspark.sql.functions import avg,sum,count


```python
from pyspark.sql.functions import avg,sum,count,round
df2.show(5)
df2.groupBy("profession").agg(round(avg("age"),2).alias("avg_age"),count("custid").alias("cust_cnt")).show(5)
```

```python
#group by in multi columns
from pyspark.sql.functions import avg,sum,count,round
df2.show(5)
df2.groupBy("profession","region").agg(round(avg("age"),2).alias("avg_age"),count("custid").alias("cust_cnt")).show(5)
```

```python
##find min age  max and avg age by prof
from pyspark.sql.functions import avg,sum,count,round,min,max
df2.groupBy("profession").agg(min("age").alias("min_age"),max("age").alias("max_age"),round(avg("age"),2).alias("avg_age")).orderBy("profession").show(5)
```
