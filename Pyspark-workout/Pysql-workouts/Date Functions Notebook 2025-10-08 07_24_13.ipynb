```python
from pyspark.sql.types import StructType, StructField, StringType,IntegerType,FloatType
from pyspark.sql.functions import col,concat,lit

cust_sch=StructType([StructField('txn_id', IntegerType(), True), StructField('txndate', StringType(), True), StructField('custid', StringType(), True), StructField('amount', FloatType(), True), StructField('Product', StringType(), True), StructField('City', StringType(), True), StructField('State', StringType(), True), StructField('Payment_mode', StringType(), True)])
df= spark.read.format("csv").option("header",False).schema(cust_sch).load("/Volumes/databricks_practice/inputdb/txns/txns.txt")
print(df.schema)

df2=df.withColumn("region",lit("USA"))
df2.printSchema()



```

```python
#converting string to date
from pyspark.sql.functions import col, to_date, date_add, date_sub, add_months, months_between, datediff, current_date


data = [
    ("2025-10-08",),
    ("2025-09-20",),
    ("2025-08-15",)
]
df4 = spark.createDataFrame(data, ["order_date_str"])
from pyspark.sql.functions import to_date,current_date,current_timestamp,date_format,date_trunc,trunc
df4=df4.withColumn("order_date",to_date(col("order_date_str")))

#Exract year,month,date,day from date column

#Current date and Timestamp functions
df4= df4.select(current_date().alias("cur_date"),current_timestamp().alias("cur_time"),date_format(current_date(),"dd-MM-yyyy").alias("cur_date_format"))

df4.show()


```

#date arithmetic 
| Function                 | Description          | Example                       |
| ------------------------ | -------------------- | ----------------------------- |
| `date_add(col, n)`       | Adds *n* days        | `2025-10-01 + 5 → 2025-10-06` |
| `date_sub(col, n)`       | Subtracts *n* days   | `2025-10-06 - 5 → 2025-10-01` |
| `add_months(col, n)`     | Adds *n* months      | `2025-10-01 + 1 → 2025-11-01` |
| `months_between(d1, d2)` | Difference in months | `6.0`                         |
| `datediff(d1, d2)`       | Difference in days   | `10`                          |
| `current_date()`         | Returns today’s date | `2025-10-08`                  |


#Date Extraction Functions:
| Function                | Description                   | Example Result         |
| ----------------------- | ----------------------------- | ---------------------- |
| `year(col)`             | Extracts year                 | 2025                   |
| `month(col)`            | Extracts month (1–12)         | 10                     |
| `dayofmonth(col)`       | Day of the month              | 8                      |
| `dayofyear(col)`        | Day number in year (1–365)    | 281                    |
| `dayofweek(col)`        | Day number in week (1=Sunday) | 4                      |
| `weekofyear(col)`       | Week number in the year       | 41                     |
| `quarter(col)`          | Quarter of year (1–4)         | 4                      |
| `date_format(col, fmt)` | Custom date formatting        | "Wednesday", "October" |
| `current_date()`        | Returns current system date   | 2025-10-08             |


```python
from pyspark.sql.functions import to_date,current_date,current_timestamp,date_format,date_trunc,trunc,year,month,dayofmonth,dayofyear,dayofweek,weekofyear,quarter,date_format
df4.withColumn("order_year",year("cur_date")).withColumn("order_month",month("cur_date")).withColumn("day_month",dayofmonth("cur_date")).withColumn("day_year",dayofyear("cur_date")).withColumn("day_week",dayofweek("cur_date")).withColumn("weekyear",weekofyear("cur_date")).withColumn("quar",quarter("cur_date")).withColumn("order_month",date_format("cur_date","dd-MM-yyyy")).show(2)
```
