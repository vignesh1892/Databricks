###WithColumn => Adding new column

```python
from pyspark.sql.types import StructType, StructField, StringType,IntegerType
from pyspark.sql.functions import col,concat,lit

cust_sch=StructType([StructField('custid', IntegerType(), True), StructField('fname', StringType(), True), StructField('lname', StringType(), True), StructField('age', IntegerType(), True), StructField('profession', StringType(), True)])
df= spark.read.format("csv").option("header",True).schema(cust_sch).load("/Volumes/databricks_practice/inputdb/customerdata/custs.txt")
print(df.schema)

df2=df.withColumn("region",lit("USA"))
df2.printSchema()
##sql 
df1=df.select("custid","fname","lname","age","profession",lit("USA").alias("region"),concat(col("fname"),lit(" "),col("lname")).alias("name"))
df1.printSchema()
df1.display()
df2.display()

```

##withColumnRenamed => rename column


```python
df2.withColumnRenamed("custid","customerid").show(5)
```

##Remove one or more column => drop()

```python
df1.drop("region").show(3)
df1.drop("region","name").show(3)
```

## na => Handling null values/missing values

```python
#drop rows where any column is having null values
df1.na.drop().show()
```

```python
#drop rows where one column is having null values
df1.na.drop(subset=["age","profession"]).show(3)
```

```python
#fill nulls rows 
df1.na.fill({"profession":"Unknown"}).show(3)
```

```python
#replace value in column
df1.na.replace({"Firefighter":"Fire Fighter","Chen":"Che"},subset=["profession","lname"]).show(3)
```

###selectExpr

```python

df1.selectExpr("custid as customer_id","region","upper(profession) as profession").limit(5).display()
```
