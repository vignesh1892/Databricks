Column-Level Security (CLS) in Databricks allows you to restrict access to specific columns within a table for different users or groups, ensuring that sensitive data is only visible to authorized personnel. This is a crucial component of a robust data governance strategy, enabling fine-grained control over your data assets.

```python
%sql
--function 
-- UDF to restrict amt
CREATE OR REPLACE FUNCTION databricks_practice.outputdb.rest_loan_amt_func(loan_amt INT)
RETURNS INT
RETURN CASE
         WHEN is_member('admin') THEN loan_amt
         ELSE  NULL -- restrict amt
       END;
```

```python
%sql
--applying the function to table
ALTER TABLE databricks_practice.outputdb.tbl_loandata
ALTER COLUMN loan_amount SET MASK databricks_practice.outputdb.rest_loan_amt_func;
```
