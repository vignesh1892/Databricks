```python
from delta.tables import DeltaTable

# Load delta table directly from path
delta_tbl = DeltaTable.forPath(spark, "/Volumes/inceptez_catalog/inputdb/employee/dept_checkpoint")

# Create checkpoint
delta_tbl.checkpoint()

print("✅ Checkpoint created successfully.")

spark.sparkContext.setCheckpointDir("/tmp/checkpoints")

df = spark.read.format("delta").load("/Volumes/inceptez_catalog/inputdb/employee/dept_checkpoint")
df.checkpoint(eager=True).write.format("delta").mode("overwrite").save("/Volumes/inceptez_catalog/inputdb/employee/dept_checkpoint")
```
