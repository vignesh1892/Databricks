```python
source_path='/Volumes/databricks_practice/inputdb/country_population'
schema_location='/Volumes/databricks_practice/inputdb/country_population/schema'
chkp_loaction='/Volumes/databricks_practice/inputdb/country_population/checkpoint'

df = spark.readStream.\
     format("cloudFiles").\
     option("cloudFiles.format","csv").\
     option("cloudFiles.schemaLocation",schema_location).\
     load(source_path)

df.writeStream.\
    format("delta").\
    option("checkpointLocation",chkp_loaction).\
    table("databricks_practice.outputdb.country_pop")
```
