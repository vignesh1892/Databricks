```python
ms_projects = [(1, 'Project1', 29498),(2, 'Project2', 32487),(3, 'Project3', 43909),(4, 'Project4', 15776),(5, 'Project5', 36268),(6, 'Project6', 41611),(7, 'Project7', 34003),(8, 'Project8', 49284),(9, 'Project9', 32341),(10, 'Project10',  47587),(11, 'Project11',  11705),(12, 'Project12',  10468),(13, 'Project13',  43238),(14, 'Project14',  30014),(15, 'Project15',  48116),(16, 'Project16',  19922),(17, 'Project17',  19061),(18, 'Project18',  10302),(19, 'Project19',  44986),(20, 'Project20',  19497)]

ms_emp_projects = [ (10592, 1),(10593, 2),(10594, 3),(10595, 4),(10596, 5),(10597, 6),(10598, 7),(10599, 8),(10600, 9),(10601, 10),(10602, 11),(10603, 12),(10604, 13),(10605, 14),(10606, 15),(10607, 16),(10608, 17),(10609, 18),(10610, 19),(10611, 20)]
from pyspark.sql.types import *
cust_schema =StructType([StructField("id",IntegerType(),True)
               ,StructField("title",StringType(),True)
               ,StructField("budget",IntegerType(),True)
               ])
cust_schema2 =StructType([StructField("emp_id",IntegerType(),True)
               ,StructField("project_id",IntegerType(),True)
               ])
prj_df=spark.createDataFrame(ms_projects,schema=cust_schema)
emp_prj_df=spark.createDataFrame(ms_emp_projects,schema=cust_schema2)

prj_df.show()
emp_prj_df.show()
```

```python
from pyspark.sql import functions as F

prj_df2 = prj_df.alias("t1").join(
    emp_prj_df.alias("t2"),
    F.col("t1.id") == F.col("t2.project_id"),
    "inner"
)

prj_df2.show()

result_df = prj_df2.groupBy(
    "title",
    "emp_id"
).agg(
    F.round(
        F.col("budget").cast("double") / F.count("emp_id"),
        0
    ).alias("budget_per_emp")
)
result_df.show()
result_df.orderBy("title").show()
```
