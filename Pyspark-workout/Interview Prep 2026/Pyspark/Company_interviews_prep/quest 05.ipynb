```python

data = [
    (2019,'Toyota','Avalon'),
    (2019,'Toyota','Camry'),
    (2020,'Toyota','Corolla'),
    (2019,'Honda','Accord'),
    (2019,'Honda','Passport'),
    (2019,'Honda','CR-V'),
    (2020,'Honda','Pilot'),
    (2019,'Honda','Civic'),
    (2020,'Chevrolet','Trailblazer'),
    (2020,'Chevrolet','Trax'),
    (2019,'Chevrolet','Traverse'),
    (2020,'Chevrolet','Blazer'),
    (2019,'Ford','Figo'),
    (2020,'Ford','Aspire'),
    (2019,'Ford','Endeavour'),
    (2020,'Jeep','Wrangler')
]

from pyspark.sql.types import StructType, StructField, IntegerType, StringType
columns = ["year", "company_name", "product_name"]
schema = StructType([
    StructField("year", IntegerType(), True),
    StructField("company_name", StringType(), True),
    StructField("product_name", StringType(), True)
])
df = spark.createDataFrame(data, schema=schema)
from pyspark.sql.functions import col
df.createOrReplaceTempView("df_v")

spark.sql(
    """
    WITH cte1 AS (
        SELECT company_name,
               SUM(CASE WHEN year = 2019 THEN 1 ELSE 0 END) AS prod_2019,
               SUM(CASE WHEN year = 2020 THEN 1 ELSE 0 END) AS prod_2020
        FROM df_v
        GROUP BY company_name
    )
    SELECT company_name,
           (prod_2020 - prod_2019 ) AS net_diff
    FROM cte1
    ORDER BY net_diff
    """
).show()

```
