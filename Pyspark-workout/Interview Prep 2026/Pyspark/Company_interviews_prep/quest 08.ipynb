```python
from pyspark.sql.types import *

nominee_schema = StructType([
    StructField("name", StringType(), True),
    StructField("amg_person_id", StringType(), True),
    StructField("top_genre", StringType(), True),
    StructField("birthday", TimestampType(), True),
    StructField("id", IntegerType(), True)
])

from datetime import datetime

nominee_data = [
    ("Jennifer Lawrence", "P562566", "Drama", datetime(1990, 8, 15), 755),
    ("Jonah Hill", "P418718", "Comedy", datetime(1983, 12, 20), 747),
    ("Anne Hathaway", "P292630", "Drama", datetime(1982, 11, 12), 744),
    ("Jennifer Hudson", "P454405", "Drama", datetime(1981, 9, 12), 742),
    ("Rinko Kikuchi", "P475244", "Drama", datetime(1981, 1, 6), 739)
]

nominee_df = spark.createDataFrame(nominee_data, nominee_schema)
nominee_df.show(truncate=False)


oscar_schema = StructType([
    StructField("year", IntegerType(), True),
    StructField("category", StringType(), True),
    StructField("nominee", StringType(), True),
    StructField("movie", StringType(), True),
    StructField("winner", IntegerType(), True),
    StructField("id", IntegerType(), True)
])

oscar_data = [
    (2008, "actress in a leading role", "Anne Hathaway", "Rachel Getting Married", 0, 77),
    (2012, "actress in a supporting role", "Anne HathawayLes", "Mis_rables", 1, 78),
    (2006, "actress in a supporting role", "Jennifer Hudson", "Dreamgirls", 1, 711),
    (2010, "actress in a leading role", "Jennifer Lawrence", "Winters Bone", 1, 717),
    (2012, "actress in a leading role", "Jennifer Lawrence", "Silver Linings Playbook", 1, 718),
    (2011, "actor in a supporting role", "Jonah Hill", "Moneyball", 0, 799),
    (2006, "actress in a supporting role", "Rinko Kikuchi", "Babel", 0, 1253)
]

oscar_df = spark.createDataFrame(oscar_data, oscar_schema)
oscar_df.show(truncate=False)
```

```python
from pyspark.sql.functions import count,col
oscar_df_w=oscar_df.filter("winner==1").groupBy("nominee").agg(count("*").alias("count_winner"))

```

```python
oscar_df_w.alias("t1").join(nominee_df.alias("t2"),on=[col("t1.nominee") == col("t2.name")],how="inner").show()
```
