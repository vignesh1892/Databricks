```python
df.createOrReplaceTempView("df_v")

spark.sql("""with avg_dp_sal as (select department,round(avg(salary),2) avg_salary
          from df_v group by department), cte2 as ( select e.id employee_id,e.department,m.id,e.salary emp_salary,m.salary manager_salary
          from df_v e inner join df_v m on e.manager_id =m.id )
          select a.department,b.employee_id employee_id,b.emp_salary,b.manager_salary ,a.avg_salary
          from avg_dp_sal a inner join cte2 b 
          on a.department =b.department""").show()
```
