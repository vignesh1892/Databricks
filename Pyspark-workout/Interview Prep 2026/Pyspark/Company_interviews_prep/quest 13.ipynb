```python
from pyspark.sql.functions import (
    col, to_date, hour, minute,
    dayofmonth, date_format
)

boi_df2 = boi_df.select(
    "transaction_id",
    to_date("time_stamp").alias("date"),
    hour("time_stamp").alias("hour"),
    minute("time_stamp").alias("minute"),
    dayofmonth("time_stamp").alias("dayofmonthtxns"),
    date_format("time_stamp", "EEE").alias("daytxns"),
    date_format("time_stamp", "HH:mm:ss").alias("timetxns")
)

filtered_df = (
    boi_df2
    .filter(~col("daytxns").isin("Sat", "Sun"))
    .filter(~col("date").isin("2022-12-25", "2022-12-26"))
    .filter((col("hour") < 9) & (col("hour") > 16))
)

filtered_df.show(truncate=False)
```
