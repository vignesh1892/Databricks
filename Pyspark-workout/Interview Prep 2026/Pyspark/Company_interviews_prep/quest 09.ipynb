```python
from pyspark.sql.types import *

inventory_schema = StructType([
    StructField("product_id", IntegerType(), True),
    StructField("product_name", StringType(), True),
    StructField("product_type", StringType(), True),
    StructField("unit", StringType(), True),
    StructField("price_unit", FloatType(), True),
    StructField("wholesale", FloatType(), True),
    StructField("current_inventory", IntegerType(), True)
])
inventory_data = [
    (1, "strawberry", "produce", "lb", 3.28, 1.77, 13),
    (2, "apple_fuji", "produce", "lb", 1.44, 0.43, 2),
    (3, "orange", "produce", "lb", 1.02, 0.37, 2),
    (4, "clementines", "produce", "lb", 1.19, 0.44, 44),
    (5, "blood_orange", "produce", "lb", 3.86, 1.66, 19)
]

inventory_df = spark.createDataFrame(inventory_data, inventory_schema)
inventory_df.show(truncate=False)
from pyspark.sql.types import *
from pyspark.sql.types import TimestampType
transaction_schema = StructType([
    StructField("transaction_id", IntegerType(), True),
    StructField("time", TimestampType(), True),
    StructField("product_id", IntegerType(), True)
])
from datetime import datetime

transaction_data = [
    (153, datetime(2016, 1, 6, 8, 57, 52), 1),
    (91,  datetime(2016, 1, 7, 12, 17, 27), 1),
    (31,  datetime(2016, 1, 5, 13, 19, 25), 1),
    (24,  datetime(2016, 1, 3, 10, 47, 44), 3),
    (4,   datetime(2016, 1, 6, 17, 57, 42), 3),
    (163, datetime(2016, 1, 3, 10, 11, 22), 3),
    (92,  datetime(2016, 1, 8, 12, 3, 20), 2),
    (32,  datetime(2016, 1, 4, 19, 37, 14), 4),
    (253, datetime(2016, 1, 6, 14, 15, 20), 5),
    (118, datetime(2016, 1, 6, 14, 27, 33), 5)
]

transaction_df = spark.createDataFrame(transaction_data, transaction_schema)
transaction_df.show(truncate=False)

```

```python
from pyspark.sql.functions import count,col
df1=transaction_df.groupBy("product_id").agg(count("transaction_id").alias("trn_cnt"))
df1.alias("t1").join(inventory_df.alias("t2"),on=col("t1.product_id")==col("t2.product_id"),how="inner").select(col("product_name"),col("trn_cnt")).show()
```
