```python
from pyspark.sql.types import *

schema = StructType([
    StructField("product_name", StringType(), True),
    StructField("mrp", StringType(), True),
    StructField("price", StringType(), True),
    StructField("pdp_url", StringType(), True),
    StructField("brand_name", StringType(), True),
    StructField("product_category", StringType(), True),
    StructField("retailer", StringType(), True),
    StructField("description", StringType(), True),
    StructField("rating", FloatType(), True),
    StructField("review_count", IntegerType(), True),
    StructField("style_attributes", StringType(), True),
    StructField("total_sizes", StringType(), True),
    StructField("available_size", StringType(), True),
    StructField("color", StringType(), True)
])
amazon_data = [
    ("ProductA", "100", "80", "url1", "BrandA", "Category1", "Amazon", "DescriptionA", 4.5, 100, "StyleA", "M,L", "M", "Red"),
    ("ProductB", "200", "180", "url2", "BrandB", "Category1", "Amazon", "DescriptionB", 4.2, 150, "StyleB", "S,M,L", "S", "Blue"),
    ("ProductC", "300", "250", "url3", "BrandC", "Category2", "Amazon", "DescriptionC", 4.8, 200, "StyleC", "L,XL", "L", "Green")
]

amazon_df = spark.createDataFrame(amazon_data, schema)
amazon_df.show(truncate=False)

macys_data = [
    ("ProductA", "100", "85", "url4", "BrandA", "Category1", "Macys", "DescriptionA", 4.5, 90, "StyleA", "M,L", "M", "Red"),
    ("ProductD", "150", "130", "url5", "BrandD", "Category3", "Macys", "DescriptionD", 4.0, 80, "StyleD", "S,M", "S", "Yellow"),
    ("ProductE", "250", "210", "url6", "BrandE", "Category4", "Macys", "DescriptionE", 3.9, 60, "StyleE", "M,L", "L", "Black")
]

macys_df = spark.createDataFrame(macys_data, schema)
macys_df.show(truncate=False)

topshop_data = [
    ("ProductB", "200", "190", "url7", "BrandB", "Category1", "TopShop", "DescriptionB", 4.1, 95, "StyleB", "S,M,L", "M", "Blue"),
    ("ProductF", "100", "90", "url8", "BrandF", "Category3", "TopShop", "DescriptionF", 3.5, 50, "StyleF", "XS,S", "S", "Pink"),
    ("ProductG", "300", "270", "url9", "BrandG", "Category5", "TopShop", "DescriptionG", 4.3, 70, "StyleG", "M,L,XL", "M", "Purple")
]

topshop_df = spark.createDataFrame(topshop_data, schema)
topshop_df.show(truncate=False)



```

```python
from pyspark.sql.functions import col

joined_df = (
    amazon_df.alias("a")
    .join(
        macys_df.alias("m"),
        (col("a.product_name") == col("m.product_name")) &
        (col("a.brand_name") == col("m.brand_name")) &
        (col("a.product_category") == col("m.product_category")),
        "left"
    )
    .join(
        topshop_df.alias("t"),
        (col("a.product_name") == col("t.product_name")) &
        (col("a.brand_name") == col("t.brand_name")) &
        (col("a.product_category") == col("t.product_category")),
        "left"
    )
).filter((col("m.product_name").isNull()) & (col("t.product_name").isNull())).select("a.product_name","a.brand_name","a.price","a.rating")

display(joined_df)
```

```python

from pyspark.sql.functions import to_timestamp

# Spark session

# ----------------------------
# Train Arrivals Data
# ----------------------------
train_arrivals_data = [
    (1, '2024-11-17 08:00'),
    (2, '2024-11-17 08:05'),
    (3, '2024-11-17 08:05'),
    (4, '2024-11-17 08:10'),
    (5, '2024-11-17 08:10'),
    (6, '2024-11-17 12:15'),
    (7, '2024-11-17 12:20'),
    (8, '2024-11-17 12:25'),
    (9, '2024-11-17 15:00'),
    (10, '2024-11-17 15:00'),
    (11, '2024-11-17 15:00'),
    (12, '2024-11-17 15:06'),
    (13, '2024-11-17 20:00'),
    (14, '2024-11-17 20:10')
]

train_arrivals_df = spark.createDataFrame(
    train_arrivals_data,
    ["train_id", "arrival_time"]
).withColumn(
    "arrival_time", to_timestamp("arrival_time")
)

# ----------------------------
# Train Departures Data
# ----------------------------
train_departures_data = [
    (1, '2024-11-17 08:15'),
    (2, '2024-11-17 08:10'),
    (3, '2024-11-17 08:20'),
    (4, '2024-11-17 08:25'),
    (5, '2024-11-17 08:20'),
    (6, '2024-11-17 13:00'),
    (7, '2024-11-17 12:25'),
    (8, '2024-11-17 12:30'),
    (9, '2024-11-17 15:05'),
    (10, '2024-11-17 15:10'),
    (11, '2024-11-17 15:15'),
    (12, '2024-11-17 15:15'),
    (13, '2024-11-17 20:15'),
    (14, '2024-11-17 20:15')
]

train_departures_df = spark.createDataFrame(
    train_departures_data,
    ["train_id", "departure_time"]
).withColumn(
    "departure_time", to_timestamp("departure_time")
)

# ----------------------------
# Validate
# ----------------------------
train_arrivals_df.show(truncate=False)
train_arrivals_df.printSchema()

train_departures_df.show(truncate=False)
train_departures_df.printSchema()
```

```python
from pyspark.sql.functions import col, lit, sum as spark_sum
from pyspark.sql.window import Window

# Arrival events
arrival_events = (
    train_arrivals_df
    .select(
        col("arrival_time").alias("event_time"),
        lit(1).alias("event_type")
    )
)

# Departure events
departure_events = (
    train_departures_df
    .select(
        col("departure_time").alias("event_time"),
        lit(-1).alias("event_type")
    )
)

# Union both → TrainTimes CTE equivalent
train_times_df = arrival_events.unionAll(departure_events)
window_spec = (
    Window
    .orderBy("event_time")
    .rowsBetween(Window.unboundedPreceding, Window.currentRow)
)

platform_count_df = train_times_df.withColumn(
    "platforms_needed",
    spark_sum("event_type").over(window_spec)
)
max_platforms = (
    platform_count_df
    .agg({"platforms_needed": "max"})
    .collect()[0][0]
)

print("Maximum platforms required:", max_platforms)

```
