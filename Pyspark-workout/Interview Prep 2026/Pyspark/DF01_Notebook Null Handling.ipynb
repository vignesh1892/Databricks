```python
#handling null in array:
from pyspark.sql import functions  as F
data = [
    (1, [10, None, 30]),
    (2, [None, 50, None]),
    (3, None)    # Entire array is NULL
]

df = spark.createDataFrame(data, ["id", "scores"])
df.show(truncate=False)

#Replace null with 0 using transform()
df.withColumn("scores_clean",F.expr("transform(scores, x -> coalesce(x,0))")).show(truncate=False)

#Remove NULL values from ARRAY using array_except
df.withColumn("null_clean",F.array_except("scores",F.array(F.lit(None)))).show(truncate=False)



```
