```python
from pyspark.sql.types import *
cust_schema= StructType([StructField('Name', StringType(), True), StructField('City', StringType(), True), StructField('Age', IntegerType(), True), StructField('proof', StringType(), True)])
df=spark.read.format("csv").option("header","true").schema(cust_schema).load("/Volumes/databricks_practice/inputdb/customerdata/customers.csv")
df.display()
```

```python
##creating delta table
df.write.saveAsTable("databricks_practice.outputdb.tbl_customer_dlta")
```

```python
%sql
create table databricks_practice.outputdb.tbl_customer_dlta2
USING DELTA
AS 
SELECT * FROM databricks_practice.outputdb.tbl_customer_dlta;


```
