```python
# List files in a directory
display(dbutils.fs.ls("dbfs:/FileStore/"))

# Copy a file from one location to another
dbutils.fs.cp("dbfs:/FileStore/test.txt", "dbfs:/FileStore/test2.txt")

# Delete a file
dbutils.fs.rm("dbfs:/FileStore/test2.txt")

# Move or rename a file
dbutils.fs.mv("dbfs:/FileStore/test.txt", "dbfs:/FileStore/test2.txt")

# Copy a folder and its contents recursively
dbutils.fs.cp("dbfs:/FileStore/test_folder/", "dbfs:/FileStore/test_folder_copy/", recurse=True)

# Delete a folder and its contents recursively
dbutils.fs.rm("dbfs:/FileStore/test_folder_copy/", recurse=True)

# Move or rename a folder recursively
dbutils.fs.mv("dbfs:/FileStore/test_folder/", "dbfs:/FileStore/test_folder_renamed/", recurse=True)

# Display contents of a file
dbutils.fs.cat("dbfs:/FileStore/test2.txt")
```
