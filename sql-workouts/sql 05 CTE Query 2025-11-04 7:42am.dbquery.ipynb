```sql
CREATE OR REPLACE TABLE databricks_practice.inputdb.tblcusts USING DELTA
AS 
SELECT 
  _c0 AS custid, 
  _c1 AS fname, 
  _c2 AS lname, 
  _c3 AS age, 
  _c4 AS prof
FROM csv.`dbfs:/Volumes/spark_workouts/customerdb/customerfile/cusdata/custs.txt`
WITH (header = "false", inferSchema = "true");


select * from databricks_practice.inputdb.tblcusts  limit 10;
select * from databricks_practice.inputdb.tbltxns  limit 10;
--get list of customers whose profession is Pilot and located in california

select c.custid,c.fname,c.lname,t.amount,t.state,t.city from databricks_practice.inputdb.tblcusts c
inner join databricks_practice.inputdb.tbltxns t
on c.custid = t.custid
where prof="Pilot" and state = "California"

```
