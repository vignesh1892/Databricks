```sql
CREATE OR REPLACE TABLE databricks_practice.inputdb.tbltxns USING DELTA
AS 
SELECT 
  _c0 AS txnid, 
  _c1 AS txndate, 
  _c2 AS custid, 
  _c3 AS amount, 
  _c4 AS category, 
  _c5 AS product, 
  _c6 AS city, 
  _c7 AS state, 
  _c8 AS paytype
FROM csv.`dbfs:/Volumes/databricks_practice/inputdb/txns`
WITH (header = "false", inferSchema = "true");


select * from  databricks_practice.inputdb.tbltxns;

describe  databricks_practice.inputdb.tbltxns;

describe formatted  databricks_practice.inputdb.tbltxns;

select count(1),state from databricks_practice.inputdb.tbltxns
group by state;



select state, count(*) from databricks_practice.inputdb.tbltxns group by state;

select state, count(*) from databricks_practice.inputdb.tbltxns group by state order by 2 desc limit 10;

select state, count(*) as txncount, round(sum(amount),2) as totalsales from databricks_practice.inputdb.tbltxns group by state order by 2 desc limit 10;

select state, category,product, count(*) as txncount, round(sum(amount),2) as totalsales from databricks_practice.inputdb.tbltxns group by state,category,product order by 5 desc;

select state, category,product, count(*) as txncount, round(sum(amount),2) as totalsales 
from databricks_practice.inputdb.tbltxns 
where (product = :v_product or :v_product = 'All')
group by state,category,product order by 5 desc;

```
