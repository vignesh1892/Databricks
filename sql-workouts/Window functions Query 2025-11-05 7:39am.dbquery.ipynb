```sql
/*
4 types of ranking functions:
  1. row_number => dedup,unique id,get record by specific category
  2.rank
  3.dense_rank
  4. ntile
  syntax:
  <function> over(partition by <col> order by <col>)
*/

with dedep
as(select *,row_number() over (partition by custid order by custid) as rn
   from databricks_practice.inputdb.tblcusts
)
select  * from dedep;

select *,row_number() over (partition by prof order by custid) as rn
   from databricks_practice.inputdb.tblcusts;
--get max amount state wise
select *
from (
select *,row_number() over(partition by state order by amount desc ) as rn from databricks_practice.inputdb.tbltxns
) where rn =1;

--get max amount city wise
select *
from (
select *,row_number() over(partition by city order by amount desc ) as rn from databricks_practice.inputdb.tbltxns
) where rn =1;

--rank function
select *
from (
select *,rank() over(partition by city order by amount desc ) as rn from databricks_practice.inputdb.tbltxns
) where rn =1;

/*The NTILE function in SQL is a window function that divides an ordered set of rows into a specified number of groups (or "buckets") and assigns a bucket number to each row. This allows for the categorization of data into approximately equal-sized portions based on a particular ordering
noumber of rows/ ntile value*/
select * from databricks_practice.inputdb.employees_sc;
SELECT
    name,
    salary,
    department,
    NTILE(2) OVER (ORDER BY salary DESC) AS salary_quartile
FROM
    databricks_practice.inputdb.employees_sc;
SELECT
    name,
    salary,
    department,
    NTILE(2) OVER ( partition by  department
ORDER BY salary DESC) AS salary_quartile
FROM
    databricks_practice.inputdb.employees_sc;




```
